```python
C = np.array([
    [3,2],
    [1,4]
])
D = np.array([
    [2,0],
    [5,1]
])
print(C+D)
print(C@D)
print(np.linalg.det(C))
print(np.linalg.inv(C))
print(C@C)
print(np.linalg.eigvals(C))

print("Det:",np.linalg.det(D))
print("Eigvals",np.linalg.eigvals(D))
```

```python
A = np.array([
    [2,1],
    [3,4]
])
b = np.array([8,18])
solution = np.linalg.solve(A,b)
print(solution)
```

```python
A = np.array([
    [4,3],
    [2,-1]
])
b = np.array([18,1])
sol = np.linalg.solve(A,b)
print(sol)
print(A@sol)
#Verification tells us that the solution is actually correct, as it should ouput [18,1]
```

```python
A = np.array([
    [1,1,1],
    [2,-1,1],
    [1,2,-1]
])
b = np.array([6,3,2])
sol = np.linalg.solve(A,b)
print(sol)
print(A@sol)
# This should return b, as 1+2+3=6, 2-2+3=3 and 1+4-3=2
```
