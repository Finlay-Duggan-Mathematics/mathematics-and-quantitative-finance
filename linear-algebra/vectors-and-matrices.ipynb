Diagonalisation makes computing high powers of matrices easier because instead of repeatedly running matrix multiplications, which are slow computationally, it can quickly compute a high power of the individual eigenvalues, and then use multiplication against the eigenvector matrices to output the final matrix in a fraction of the time of simply multiplying continuously

```python
A = np.array([
    [4,1],
    [2,3]
])
eigvals, P = np.linalg.eig(A)
D = np.diag(eigvals)
Pinv = np.linalg.inv(P)
x, y = np.linalg.matrix_power(A,100), np.linalg.matrix_power(D,100)
z = P@y@Pinv
print(z,"\n\n",x,"\n\n",np.allclose(z,x),"\n\n")

Pinv = np.linalg.inv(P)
x, y = np.linalg.matrix_power(A,500), np.linalg.matrix_power(D,500)
z = P@y@Pinv
print(z,"\n\n",x,"\n\n",np.allclose(z,x),"\n\n")
```
